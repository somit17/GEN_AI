# Static prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm_model = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
prompts = [
            ("system","You are a python developer"), #(type,) "system","ai","user" ---> "PROMPT"
            ("user","How to sort the array.")
           ]
res = llm_model.invoke(prompts)

print(res.content)
```
