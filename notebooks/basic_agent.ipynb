```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0,
    max_tokens=None,
    reasoning_format="parsed",
    timeout=None,
    max_retries=2,
    streaming=True
    # other params...
)
```

######   REACT = REASONING + ACTION

```python
from langchain.tools import tool
```

```python
@tool("add")
def add_numbers(num1:int,num2:int):
    """
        It will return sum of two numbers

        Args:
            num1 : Number One
            num2 : Number Two
    """
    return num1 + num2
```

```python
@tool("multiply")
def multiply_numbers(num1:int,num2:int):
    """
        It will return product of two numbers

        Args:
            num1 : Number One
            num2 : Number Two
    """
    return num1 * num2
```

```python
add_numbers.invoke({"num1":10,"num2":20})
```

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers,multiply_numbers],
    system_prompt="You are math teacher and always use tool for calculation"
)
```

```python
respone = agent.invoke({"messages":[{"role":"user","content":"What is 2 + 3 * 30"}]})
```
