```python
from pydantic import BaseModel,Field,StrictInt
```

```python
class User(BaseModel):
    name:str=''
    age:StrictInt= 0
```

```python
user_2 = User(name='A',age=23)
print(user_2)
```

```python
# Field Validation
```

```python
class Product(BaseModel):
    title:str=Field(max_length=10)
    description:str=Field(min_length=15)
    price:int=Field(gt=0,le=100)
```

```python
earphone = Product(title='Earphone',description='This is amazing earphone from Apple',price=20)
mobile =  Product(title='SAM FE23',description='This is amazing phone from Samsung',price=80)
print(earphone)
```

```python
class Order(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    product:Product
```

```python
order_1 = Order(id="1A22343",product=earphone)
print(order_1)
```

```python
order_1.product
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items: List[Product]
    comment:str=Field(default="No Comment")
```

```python
cart = Carts(id="A34tB23",items=[earphone,mobile])
```

```python
print(cart )
```

```python
for product in cart.items:
    print(f"Product Name - {product.title}")
```

```python
### Custom Validator
```

```python
common_validator = Annotated[List[Product],Field(default=[])]
```

```python
class CartV2(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items: common_validator
    comment:str=Field(default="No Comment")
```

```python
class OrderV2(BaseModel):
    products:common_validator
```

```python
ord = OrderV2(products=cart.items)
```
