```python
from langchain_community.document_loaders import TextLoader,PyPDFLoader,CSVLoader,WebBaseLoader,WikipediaLoader
```

### Text DOC LOADER

```python
texts = TextLoader('../../data/speech.txt')
docs = texts.load()
print(docs)
```

```python
pdf_doc_loader = PyPDFLoader('../../data/medical_report.pdf')
pdf = pdf_doc_loader.load()
print(len(pdf))
```

```python
pdf_result = pdf[0]
print(pdf_result.page_content)
```

#### CSV File LOADER

```python
csv_loader = CSVLoader('../../data/employee_records.csv')

csv = csv_loader.load_and_split()
print(len(csv ))
```

#### Web Page Loader

```python
web_loader = WebBaseLoader('https://docs.langchain.com/oss/python/integrations/document_loaders/index#webpages')

web = web_loader.load()
for each_page in web:
    print(each_page.page_content)
```

### WIKIPEDIA LOADER

```python
wiki_loader=WikipediaLoader(query="GEN AI",load_max_docs=2)
wiki = wiki_loader.load()
```

```python
for each_page in wiki:
    print(each_page.page_content)
```
